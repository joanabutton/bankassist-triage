# BankAssist Triage — Business Problem & Use Case

## 1. Project Overview

BankAssist Triage is an academic machine-learning prototype designed to support the triage and routing of short digital-banking customer queries.
The system classifies each incoming message into one of 10 supported banking intents and uses the predicted intent to suggest the appropriate support destination. It also returns a model confidence score so that uncertain cases can be flagged for operational human review.
The objective is not to replace customer-support teams, but to support the first stage of query handling by making routing faster and more consistent.

## 2. Business Problem

Banks receive large numbers of short customer messages that need to be directed quickly to the appropriate support team.
When this triage is performed manually, routing can be:
- slow;
- inconsistent between agents;
- dependent on individual interpretation;
- a source of unnecessary transfers between departments.
For customers dealing with issues involving their money, repeated transfers or long queues can increase frustration and extend the time required to reach the appropriate support team.
The business problem can therefore be summarised as:
How can machine learning support faster and more consistent first-level routing of banking support messages while preserving human oversight for uncertain cases?

## 3. Proposed Use Case

BankAssist addresses this problem as an intent-classification and routing-support system.
For each new customer message, the system follows this flow:

Customer Message
       ↓
TF-IDF Transformation
       ↓
Intent Classifier
       ↓
Predicted Banking Intent
       ↓
Routing Destination
       ↓
Confidence / Operational Review Decision

The classifier does not generate a banking answer and does not perform actions on a customer's account.
Its role is to support the triage decision: identifying the likely intent of the message and suggesting where it should be routed.
BankAssist is an academic demonstration of support triage and should not be used to make banking decisions or take action on customer accounts without human oversight.

## 4. Why Smooth Triage Matters — Key External Findings

External banking research provides useful context for why reducing friction at the first point of contact may matter. These findings do not prove that better triage alone will improve retention, trust or satisfaction, but they show that difficult and slow service experiences are meaningful problems in banking.

| Finding | Relevance to BankAssist |
| --- | --- |
| **12% call abandonment at Tier I banks and nearly 18% at Tier II banks globally**; 55% of banks reported first-contact resolution below 70% (Capgemini, 2024) | Suggests that abandonment and repeated contact are material operational issues in banking, making smoother first-level handling worth testing |
| **61% of banking customers who contacted a human agent did so because they were unhappy with a chatbot's resolution** (Capgemini, 2024) | Shows that unsuccessful automated interactions can create additional customer effort and escalation to human support |
| Average problem-resolution time at U.S. direct banks increased from **1.9 to 2.6 days**, while satisfaction with ease of resolution fell **67 points** (J.D. Power, 2024) | Suggests that slower problem resolution is associated with poorer customer experience |
| Among U.S. retail-bank customers likely to switch banks, **26% cited poor service experience** as a reason (J.D. Power, 2024) | Indicates that poor service can represent a customer-retention risk |
| Banking customer trust was **8.4× higher after an easy experience**; **65%** were likely to recommend the bank after an easy experience versus **5%** after a difficult one (Qualtrics XM Institute, 2024) | Shows a strong association between lower customer effort and trust / advocacy |

### Business hypothesis for BankAssist

A smoother first-level routing process may help reduce avoidable transfers, waiting and customer effort.
External research suggests that difficult or slow service experiences are associated with lower satisfaction, trust and loyalty.

BankAssist therefore targets an early stage of the support journey where friction may be reduced, but a real-world pilot would be required to test whether improved routing actually changes downstream outcomes.

## 5. Business Value

The proposed solution is intended to support several improvements in first-level customer-service operations:

- Faster first-level triage
- More consistent routing
- Fewer unnecessary transfers
- Reduced handling time
- Reduced customer frustration

Human oversight remains part of the design: BankAssist supports rather than replaces support teams.
The main value proposition is therefore not simply classification accuracy. It is the ability to convert a short customer message into a consistent and traceable routing suggestion.

## 6. Scope of the Classifier

The project deliberately uses a small and manageable classification problem.
The source dataset contains:
- 1,000 customer messages
- 10 banking intent classes
- short English-language banking-support requests

The limited scale and linguistic complexity allow the project to focus on the complete ML/MLOps workflow rather than on large-scale data engineering.

The supported intents are:

apple_pay_or_google_pay;
beneficiary_not_allowed;
cancel_transfer;
card_about_to_expire;
card_payment_fee_charged;
change_pin;
getting_spare_card;
lost_or_stolen_phone;
supported_cards_and_currencies;
visa_or_mastercard;

Messages whose true intent falls outside these supported categories are outside the classifier's intended scope.

## 7. From Model Performance to Business Success

Good classification performance is necessary, but it is not the final business objective.
The business objective is to help a customer reach the correct support team faster, with fewer unnecessary transfers and without removing human oversight.

For that reason, BankAssist should be evaluated at several levels:

| Dimension | Example metrics | Why it matters |
| --- | --- | --- |
| Model quality | Macro F1, per-intent precision/recall, confusion matrix | Shows whether the classifier performs consistently across the 10 supported intents |
| Routing quality | Correct first-route rate, re-route / transfer rate | Measures whether customers are sent to the correct operational destination on the first attempt |
| Customer journey | Time to correct support team, first-level handling time | Connects model performance directly to speed and customer effort |
| Human oversight | Human-review rate, manual-correction rate, automation coverage | Shows how often the system needs intervention and whether the confidence policy is useful |
| Operational performance | End-to-end latency, API error rate, availability, throughput | Shows whether the system is fast and reliable enough to support a frictionless experience |

An important distinction is that an **intent-classification error is not always the same as a routing failure**.
Several supported intents can map to the same operational destination. A message could therefore be assigned the wrong fine-grained intent but still reach the correct support team.
For business evaluation, routing-level performance should therefore be monitored alongside intent-level model metrics.

### Business and operational metrics for a real-world pilot

A larger pilot should track:

- **Correct first-route rate:** percentage of messages that reach the correct support destination without being re-routed.
- **Re-route / transfer rate:** percentage of cases moved to another team after the initial routing decision.
- **Time to correct support team:** elapsed time from the customer's first message to arrival at the appropriate queue.
- **First-level handling time:** time spent on the initial triage step.
- **Manual correction rate:** percentage of BankAssist routing suggestions that support staff change.
- **Human-review rate:** proportion of messages flagged for operational human review.
- **Automation coverage:** proportion of messages routed without human review under the chosen confidence policy.
- **Macro F1 and per-intent precision/recall:** measures of classification quality across supported intents.
- **Confidence distribution:** helps assess whether the model is consistently confident or frequently uncertain.
- **Performance on reviewed production examples:** once verified feedback exists, this provides a more realistic measure than relying indefinitely on the original test set.

These metrics should be interpreted together. For example, maximizing automation coverage is not useful if it increases incorrect routing. Likewise, sending almost every message to human review could reduce routing risk but would undermine the efficiency objective.

## 8. Latency and Reliability

Latency is an important operational indicator because the purpose of BankAssist is fast, low-friction triage.
Even a highly accurate classifier would provide limited business value if customers or support systems had to wait noticeably for every routing decision.

For a pilot, two types of latency should be distinguished:

- **Model inference latency:** time required by the TF-IDF + Logistic Regression pipeline to produce a prediction.
- **End-to-end latency:** time from the frontend request to the completed API response, including network and hosting overhead.

The second measure is more closely connected to the customer's actual experience.

Rather than reporting only an average, a production-oriented pilot should monitor:
- median (p50) latency;
- p95 latency, to identify slower requests;
- API error rate;
- service availability;
- request volume / throughput.

No latency service-level target has been established in the current project. A future pilot should first measure a baseline and then define an acceptable operational target based on the intended customer journey.

## 9. Pilot Success Framework

For this academic prototype, the existing offline evaluation demonstrates that the approach is technically viable on the reviewed dataset.
A real-world pilot, however, should be judged against business, model, operational and safeguard metrics together.

A practical pilot scorecard could contain:

| Dimension | Metric | Desired direction |
| --- | --- | --- |
| Business | Correct first-route rate | Higher |
| Business | Re-route / transfer rate | Lower |
| Business | Time to correct support team | Lower |
| Efficiency | First-level handling time | Lower |
| Human oversight | Manual correction rate | Lower |
| Human oversight | Human-review rate | Balanced against routing risk |
| Model | Macro F1 / per-intent recall | Stable or higher |
| Operations | p50 and p95 end-to-end latency | Low and stable |
| Operations | API error rate | Lower |
| Reliability | Availability | Higher |
| Risk | Out-of-scope / unsupported-query rate | Monitored |
| Risk | Performance drift across message types | Stable |

The project does not currently have measured production baselines for most of these indicators.
They should therefore be treated as proposed pilot KPIs rather than achieved results.

The central evaluation question is:

> **Does BankAssist help customers reach the correct support team faster and with fewer unnecessary transfers, while maintaining acceptable routing quality, latency and human oversight?**
